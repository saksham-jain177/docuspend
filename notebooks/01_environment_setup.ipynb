```python
# Step 2: Install core dependencies
import subprocess
import sys

dependencies = [
    "paddlepaddle-gpu>=3.0.0",
    "paddleocr>=2.8.1",
    "numpy>=1.24.0",
    "opencv-python>=4.8.0",
    "pillow>=10.0.0",
    "scipy>=1.11.0",
    "scikit-image>=0.21.0",
    "datasets>=2.14.0",
    "huggingface-hub>=0.17.0",
    "pyyaml>=6.0",
    "tqdm>=4.66.0",
    "pandas>=2.0.0",
    "matplotlib>=3.8.0",
    "seaborn>=0.13.0",
    "tensorboard>=2.14.0",
]

print("Installing dependencies...")
for dep in dependencies:
    print(f"  Installing {dep}...")
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", dep])

print("\n✓ All dependencies installed")
```

```python
# Step 3: Verify imports
print("Verifying imports...")

try:
    from paddleocr import PaddleOCR
    print("✓ PaddleOCR importable")
except ImportError as e:
    print(f"✗ PaddleOCR import failed: {e}")

try:
    import cv2
    import numpy as np
    import PIL
    import pandas as pd
    import yaml
    import matplotlib
    print("✓ Image processing libraries available")
except ImportError as e:
    print(f"✗ Import failed: {e}")

try:
    from datasets import load_dataset
    print("✓ Hugging Face datasets available")
except ImportError as e:
    print(f"✗ HF datasets import failed: {e}")

print("\nLibrary versions:")
import cv2
print(f"  OpenCV: {cv2.__version__}")
import numpy
print(f"  NumPy: {numpy.__version__}")
import PIL
print(f"  Pillow: {PIL.__version__}")
```

```python
# Step 4: Create project directory structure
import os
from pathlib import Path

base_dir = Path("/content/docuspend") if os.path.exists("/content") else Path("./docuspend")
print(f"Creating project structure in: {base_dir}")

directories = [
    "data/raw/train", "data/raw/val", "data/raw/test",
    "data/processed/train", "data/processed/val", "data/processed/test",
    "data/annotations/train", "data/annotations/val", "data/annotations/test",
    "data/splits",
    "models/pretrained", "models/checkpoints", "models/final",
    "configs",
    "outputs/predictions/individual", "outputs/visualizations", "outputs/logs/tensorboard",
    "notebooks", "src"
]

created_count = 0
for directory in directories:
    dir_path = base_dir / directory
    dir_path.mkdir(parents=True, exist_ok=True)
    created_count += 1

print(f"✓ Created {created_count} directories")
print(f"✓ Project structure ready at: {base_dir}")
```
